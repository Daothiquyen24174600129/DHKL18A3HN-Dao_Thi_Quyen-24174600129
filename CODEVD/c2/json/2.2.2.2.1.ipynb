```python
import json 

# Đối tượng Python cần chuyển đổi
data = {
    "name": "A",
    "age": 35,
    "company": "Đất Việt"
}

print(type(data))  # data là đối tượng Python kiểu dict

# Chuyển đối tượng Python thành chuỗi JSON
json_string = json.dumps(data, ensure_ascii=False)  # Hàm dumps() chuyển dict thành chuỗi JSON

print(json_string)            # In chuỗi JSON
print(type(json_string))      # json_string là kiểu str, không phải dict nữa
```
