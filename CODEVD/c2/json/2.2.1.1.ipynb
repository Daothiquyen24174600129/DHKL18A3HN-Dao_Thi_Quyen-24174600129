```python
import json 

def complex_handler(obj):  # Hàm xử lý số phức
    if isinstance(obj, complex):
        return [obj.real, obj.imag]  # Trả về danh sách gồm phần thực và phần ảo
    raise TypeError(repr(obj) + " không thể mã hóa JSON")  # Nếu không phải số phức, báo lỗi

# Ứng dụng hàm complex_handler để mã hóa số phức
data = {"z": 2 + 3j}
complex_obj = json.dumps(data, default=complex_handler)  # Sử dụng đối số default để chỉ định hàm xử lý
print(complex_obj)
```
