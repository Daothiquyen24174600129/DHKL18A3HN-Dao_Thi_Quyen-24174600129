```python
import tkinter as tk
from tkinter import Spinbox

def on_spinbox_changed():
    value = spinbox.get()
    label2.config(text="Giá trị đã chọn: " + value)

# Tạo cửa sổ
root = tk.Tk()
root.title("SpinBox Example")
root.geometry('300x100')

# Tạo Spinbox
spinbox = Spinbox(root, from_=1, to=10, increment=1, width=10, command=on_spinbox_changed)
spinbox.pack(pady=10)

# Tạo Label hiển thị giá trị
label2 = tk.Label(root, text="Giá trị đã chọn: 1")
label2.pack()

# Chạy cửa sổ
root.mainloop()
```
