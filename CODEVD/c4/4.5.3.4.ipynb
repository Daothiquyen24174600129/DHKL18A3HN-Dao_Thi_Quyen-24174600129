```python
import pandas as pd
import numpy as np

print("— 6. Chuẩn bị DataFrame chứa dữ liệu:")

# Dữ liệu mẫu có chứa NaN
data = {
    'A': [1.0, 2.0, np.nan, 4.0],
    'B': [np.nan, 6.8, 2.8, 8.01],
    'C': [10, 11, 12, 13]
}

# Tạo DataFrame
df_demo = pd.DataFrame(data)

print("DataFrame gốc (df_demo):\n", df_demo)
```

```python
print("\n— 7. Demo isnull() và Lọc -")

# Tạo mask kiểm tra giá trị NaN
mask_na = df_demo.isnull()

# In kết quả
print(f"Kết quả isnull():\n{mask_na}")
```

```python
# Lọc các hàng không có bất kỳ giá trị NA nào (chỉ giữ lại hàng sạch)
df_sach = df_demo[df_demo.notnull().all(axis=1)]
print("\nDataFrame sau khi lọc các hàng sạch:\n", df_sach)
```

```python

# Tạo DataFrame mẫu
df_all_na_col = pd.DataFrame({
    'A': [1, 2, np.nan,4],
    'B': [np.nan, 6, 7,8],
    'C': [9,10,11,12],  # Toàn bộ là NaN
    'D': [5,6,np.nan,8]
})

# Xóa các cột mà toàn bộ giá trị là NaN
df_xoa_cot = df_all_na_col.dropna(axis='columns', how='all')
print(f"\nKết quả dropna(axis='columns', how='all'):\n{df_xoa_cot}")
```

```python
# Xóa các cột (columns) mà BẤT KỲ (how='any') giá trị nào là NA
df_xoa_cot_any = df_all_na_col.dropna(axis='columns', how='any')  
#print(f"\nKết quả dropna(axis='columns', how='any'):\n{df_xoa_cot_any}")  # Cột A, B, D và E bị xóa

df_xoa_cot_any
```
