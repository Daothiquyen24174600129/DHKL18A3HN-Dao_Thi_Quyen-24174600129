```python
import pandas as pd

# Tạo Series về diện tích
area_dict = {
    'Hà Nội': 3358.9,
    'Quảng Ninh': 6177.7,
    'Nam Định': 1668.0,
    'Đà Nẵng': 1284.9,
    'TP.Hồ Chí Minh': 2061.0
}
area = pd.Series(area_dict)

# Tạo Series về dân số
population_dict = {
    'Hà Nội': 8250000,
    'Quảng Ninh': 1400000,
    'Nam Định': 1800000,
    'Đà Nẵng': 1200000,
    'TP.Hồ Chí Minh': 9100000
}
population = pd.Series(population_dict)

# Tạo DataFrame từ 2 Series
VietNam = pd.DataFrame({'Dân số': population, 'Diện tích': area})
print(VietNam)
```
