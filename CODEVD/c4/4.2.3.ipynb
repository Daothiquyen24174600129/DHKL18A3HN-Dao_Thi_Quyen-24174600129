```python
import pandas as pd
import numpy as np

# Tạo Series
s = pd.Series({'a': 1.0, 'b': 2.0, 'c': 3.0})
print("Series:\n", s)

# Tạo DataFrame
arr = np.arange(1, 10).reshape(3, 3)
df = pd.DataFrame(arr, columns=['col1', 'col2', 'col3'])  # <-- sửa 'colums' thành 'columns'
print("DataFrame:\n", df)  # <-- sửa '\DataFrame' thành 'DataFrame'
```
