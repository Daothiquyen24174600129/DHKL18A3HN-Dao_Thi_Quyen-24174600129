```python
import numpy as np
import pandas as pd
arr_none = np.array([1, 2, None, 3])
print("1. Mang co None (dtype=object):")
print(arr_none)
```

```python
type(arr_none)
```

```python
print("\n2. Loi khi tinh mean() tren mang object chua none: ")
try:
    arr_none.mean()
except TypeError as e:
    print(f"--> loi: {e}")
```

```python


print("\n3. Minh họa lỗi khi tính sum() trên mảng object thuần:")

# Tạo mảng object, chú ý các dấu ngoặc phải đúng
arr_pure_obj = np.array([1, 2, 3, '4'], dtype=object)

try:
    # Ngay cả khi không có None, object array vẫn gây lỗi nếu có kiểu khác số
    arr_pure_obj.sum()
except TypeError as e:
    print(f"→ Lỗi: {e} (do có chuỗi '4')")
```

```python


for dtype in ['object', 'int']:
    print("dtype =", dtype)
    %timeit np.arange(1e6, dtype=dtype).sum()
    print()
```
