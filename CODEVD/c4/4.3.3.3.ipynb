```python
import numpy as np
import pandas as pd  # ← thêm dòng này

# Định nghĩa kiểu dữ liệu có cấu trúc
dtype = [('Họ tên', 'U10'), ('Tuổi', 'i4'), ('Chiều cao', 'f4')]

# Tạo mảng dữ liệu có cấu trúc
data = np.array([
    ('Minh', 22, 165.5),
    ('Lan', 40, 160.0),
    ('Linh', 31, 165.0),
    ('Lương', 18, 166.5),
    ('Thanh', 50, 170.0)
], dtype=dtype)

# Tạo DataFrame từ Structured Array
df = pd.DataFrame(data)

# In ra DataFrame
print(df)
```
