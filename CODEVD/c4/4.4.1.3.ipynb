```python
import pandas as pd 
data = pd.Series([1, 2, 3, 4], 
                index=['a','b','c','d']) 
data 
```

```python
#Array slicing (explicit index)  
#lấy các phần tử trong  khoảng index từ a -> c (do người dùng định nghĩa) 
data['a':'c'] 
```

```python
#Array slicing (implicit  index)  
# lấy các phần tử có index là 0 và 1 không lấy 2. 
data[0:2] 
```
