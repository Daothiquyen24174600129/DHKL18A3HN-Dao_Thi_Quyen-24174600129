```python
import pandas as pd

# Dữ liệu minh họa
vn_pop_df = pd.DataFrame({
    2015: [7000000, 1800000, 3500000, 1100000, 8000000],
    2020: [7200000, 1900000, 3600000, 1200000, 8500000]
}, index=['Hà Nội', 'Nam Định', 'Thanh Hóa', 'Đà Nẵng', 'TP.HCM'])

# Số người có việc làm
co_viec_2015 = [3752320, 1080750, 1512340, 512340, 4920000]
co_viec_2020 = [4042454, 1011049, 1600000, 600000, 5200000]

# Tạo DataFrame có MultiIndex cho cột
vn_pop_df_expanded = pd.DataFrame({
    ('Tổng dân số', 2015): vn_pop_df[2015],
    ('Tổng dân số', 2020): vn_pop_df[2020],
    ('Có việc', 2015): co_viec_2015,
    ('Có việc', 2020): co_viec_2020
}, index=vn_pop_df.index)

# Tạo MultiIndex cho cột
vn_pop_df_expanded.columns = pd.MultiIndex.from_tuples(vn_pop_df_expanded.columns)

# Hiển thị kết quả
print(vn_pop_df_expanded)
```
