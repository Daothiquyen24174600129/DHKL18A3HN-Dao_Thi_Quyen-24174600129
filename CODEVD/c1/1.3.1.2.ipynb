```python
lass Dog: 
    def __init__(self, name, size, age, color):
        self._name = name      
        self._size = size    
        self.__age = age       
        self.__color = color 

    @property   
    def age(self):
        print("Getter for age called")
        if not hasattr(self, "_Dog__age"): 
            raise AttributeError("Age has been deleted or not set")
        return self.__age       

    @age.setter
    def age(self, value):   
        print("Setter for age called")
        if value < 0:
            raise ValueError("Age cannot be negative")
        self.__age = value  

    @age.deleter  
    def age(self):
        print("Deleter for age called")
        if not hasattr(self, "_Dog__age"): 
            raise AttributeError("Age has already been deleted or not set")
        del self.__age   
dog = Dog("Bull", "large", 2, "yellow")

print(f"Bull's age is: {dog.age}")

```
