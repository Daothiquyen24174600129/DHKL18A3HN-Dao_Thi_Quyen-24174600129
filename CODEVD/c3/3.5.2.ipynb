```python
import numpy as np 
visitors = np.array([10200, 9875, 13567,7900,11200, 8871, 12567]) 
```

```python
sum = 0 
for i in visitors: 
    if (i>10000): 
        sum+=i 
print(sum) 
```

```python
mask = visitors > 10000 
mask
```

```python
#Áp dụng mask để lọc các phần  tử thỏa mãn điều kiện (>10000) 
visitors[mask] 
```

```python
#Tính tổng mảng các phần tử được lọc ra 
visitors[mask].sum() 
```
