```python
import numpy as np
arr = np.arange(16).reshape(4, 4) 
print("Mảng gốc ban đầu:\n arr = ", arr) 
  
#Lấy các phần tử arr[3, 1], arr[3, 2] và arr[3, 3] 
sub_arr = arr[3, [1, 2, 3]] 
print("Mảng con: sub_arr = ", sub_arr)
```

```python
# 3.6.2.2
# Lấy tất cả các phần tử nằm tại hàng >= 1 và các cột = 1, 2, 3 tương ứng 
sub_arr = arr[1:, [1, 2, 3]] 
print("Mảng con sub_arr:\n", sub_arr) 
```
