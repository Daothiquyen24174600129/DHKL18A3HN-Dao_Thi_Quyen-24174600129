```python
import numpy as np
x = np.array([1, 2, 3]) 
y = np.array([4, 5, 6]) 
np.concatenate([x,y]) #Kết hợp ma trận x, y 
```

```python
np.concatenate([x,y], axis=1)
```

```python
import numpy as np 
a = np.array([[1, 2], 
              [3, 4]]) 
b = np.array([[5, 6], 
              [7, 8]]) 
# Kết hợp theo chiều đầu tiên (trục dọc/ trục 0) của mảng 
np.concatenate((a, b))  
```

```python
np.concatenate([a,b], axis=1)
```
