```python
import numpy as np 
 
# Định nghĩa kiểu dữ liệu cho một Structured Array 
dtype = [('name', 'U20'), ('age', int), ('height', float)] 
 
# Tạo một Structured Array 
data = np.array([('Minh', 22, 165.5), 
                 ('Lan', 40, 160.0), 
                 ('Linh', 31, 165.0), 
                 ('Lương',18,166.5), 
                 ('Thanh',50, 170.0)], dtype=dtype) 
# Truy cập các trường bằng tên 
print(data['name'])    # In ra mảng 'name' 
print(data['age'])     # In ra mảng 'age' 
print(data['height'])  # In ra mảng 'height'
```
